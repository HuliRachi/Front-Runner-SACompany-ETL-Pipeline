```python
dbutils.widgets.text("catalog", "dev", "Catalog")
catalog = dbutils.widgets.get("catalog")
print(f"Running setup for catalog: {catalog}")

spark.sql(f"CREATE CATALOG IF NOT EXISTS {catalog}")
display(spark.sql(f"SHOW CATALOGS LIKE '{catalog}'"))

spark.sql(f"CREATE SCHEMA IF NOT EXISTS {catalog}.frontrunner")
display(spark.sql(f"SHOW SCHEMAS IN {catalog} LIKE 'frontrunner' "))

spark.sql(f"CREATE VOLUME IF NOT EXISTS {catalog}.frontrunner.landing")

landing_root = f"/Volumes/{catalog}/frontrunner/landing"

subfolders = [
    "clickstream",
    "products",
    "orders",
    "order_items",
    "inventory",
    "customers_cdc"
]
for folder in subfolders:
    dbutils.fs.mkdirs(f"{landing_root}/{folder}")
    print(f"Created: {landing_root}/{folder}")

display(dbutils.fs.ls(landing_root))
```
